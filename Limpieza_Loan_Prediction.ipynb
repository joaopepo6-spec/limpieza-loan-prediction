```python
import pandas as pd
df = pd.read_csv("train_u6lujuX_CVtuZ9i.csv")
print(df.head())
```

```python
print(f"Dimensiones del dataset: {df.shape}")
df.info()
```

```python
print("Valores nulos por columna:")
print(df.isnull().sum())
```

```python
print(df.describe())
```

```python
print("Filas duplicadas:", df.duplicated().sum())
```

```python
for col in ['Gender', 'Married', 'Self_Employed']:
    moda = df[col].mode()[0]
    df[col] = df[col].fillna(moda)
    print(f"{col} -> nulos rellenados con: {moda}")
```

```python
moda_dependents = df['Dependents'].mode()[0]
df['Dependents'] = df['Dependents'].fillna(moda_dependents)
df['Dependents'] = df['Dependents'].replace('3+', '3')
df['Dependents'] = df['Dependents'].astype(int)

print(df['Dependents'].unique())
print(df['Dependents'].dtype)
```

```python
mediana_loan = df['LoanAmount'].median()
df['LoanAmount'] = df['LoanAmount'].fillna(mediana_loan)
print("LoanAmount rellenado con la mediana:", mediana_loan)
print("Nulos restantes en LoanAmount:", df['LoanAmount'].isnull().sum())
```

```python
for col in ['Loan_Amount_Term', 'Credit_History']:
    moda = df[col].mode()[0]
    df[col] = df[col].fillna(moda)
    df[col] = df[col].astype(int)
    print(f"{col} -> rellenado con moda {moda}, tipo ahora: {df[col].dtype}")
```

```python
print("Nulos restantes en todo el dataset:")
print(df.isnull().sum())
```

```python
print("Monto promedio del prestamo por zona:")
print(df.groupby('Property_Area')['LoanAmount'].mean())
```

```python
print("Cantidad de prestamos por estado civil y resultado:")
print(df.groupby(['Married', 'Loan_Status']).size())
```

```python
print("Ingreso promedio del solicitante por nivel educativo:")
print(df.groupby('Education')['ApplicantIncome'].mean())
```
